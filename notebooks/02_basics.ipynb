## 1. 变量与赋值

**变量就是给一个值起个名字**，之后可以用名字来引用它。

```
名字 = 值
```

等号 `=` 在编程里**不是数学的「相等」**，而是「把右边的东西，装进左边这个名字里」。
读的时候要从右往左读：`age = 30` 读作「把 30 装进 age」。

### 命名规范：snake_case

Python 社区的约定是**全小写 + 下划线分词**，叫 snake_case（蛇形命名）：

| ✅ 推荐 | ❌ 不推荐 |
|---|---|
| `user_name` | `userName`（这是 JS/Java 风格） |
| `total_amount` | `TotalAmount`（这是类名的风格） |
| `day_count` | `d`、`x`、`a1`（看不出含义） |
| `is_finished` | `flag`、`temp`（太模糊） |

**全大写的名字**（如 `TOTAL_DAYS = 84`）约定用来表示「常量」——定义之后不再改变的值。

### 三个新手常踩的坑

````python
# ── 坑 1：变量必须先赋值才能用 ──
# 下面这行如果取消注释，会报 NameError: name 'price' is not defined
# print(price)

price = 100          # 现在 price 才存在
print(price)


# ── 坑 2：赋值是有方向的，不能反着写 ──
a = 5
b = a                # 把 a 的值(5)复制给 b
a = 99               # 之后改 a，不影响 b
print(f"a = {a}, b = {b}")   # b 还是 5，不是 99


# ── 坑 3：可以同时给多个变量赋值 ──
x, y = 1, 2
print(f"x = {x}, y = {y}")

x, y = y, x          # 交换两个变量，Python 的写法非常简洁
print(f"交换后 x = {x}, y = {y}")
````

## 2. 四种基本类型

Python 里最常用的四种类型：

| 类型 | 中文名 | 例子 | 说明 |
|---|---|---|---|
| `int` | 整数 | `42`, `-7`, `0` | 没有小数点 |
| `float` | 浮点数 | `3.14`, `-0.5`, `1.0` | 有小数点。**注意 `1.0` 是 float 不是 int** |
| `str` | 字符串 | `"你好"`, `'abc'` | 文本。单引号双引号完全等价 |
| `bool` | 布尔值 | `True`, `False` | 只有这两个值，**首字母必须大写** |

用 `type()` 可以查任何值的类型。

````python
print(type(42))          # <class 'int'>
print(type(3.14))         # <class 'float'>
print(type(1.0))          # 有小数点就是 float，哪怕值是整数
print(type("你好"))        # <class 'str'>
print(type(True))         # <class 'bool'>

print()
# ── 一个反直觉的事实：bool 其实是 int 的子类 ──
print(f"True + True = {True + True}")     # 2 ！True 当 1 用，False 当 0 用
print(f"True * 5  = {True * 5}")          # 5
````

### 类型转换

用类型名当函数就能转换：`int()` `float()` `str()` `bool()`

**这里藏着 Python 最经典的一个坑**，你今天必须记住：

````python
# ── 坑：字符串和数字不能直接相加 ──
age = 30
# print("我今年" + age + "岁")     # ← 取消注释会报 TypeError

# 必须先把数字转成字符串
print("我今年" + str(age) + "岁")

# 或者用 f-string（推荐，后面会讲）
print(f"我今年{age}岁")


print()
# ── 坑：int() 是「砍掉」小数，不是四舍五入 ──
print(f"int(3.99) = {int(3.99)}")     # 3，不是 4！昨天你已经踩过这个
print(f"int(3.01) = {int(3.01)}")     # 3
print(f"int(-3.5) = {int(-3.5)}")     # -3，向零方向砍

# 真要四舍五入，用 round()
print(f"round(3.99) = {round(3.99)}")   # 4


print()
# ── 坑：input() 拿到的永远是字符串 ──
# 这是新手最容易出的 bug，没有之一。
user_typed = "18"                      # 假装这是 input() 的结果
print(f"type(user_typed)      = {type(user_typed)}")        # <class 'str'>，不是 int！
print(f"int(user_typed) + 1   = {int(user_typed) + 1}")     # 必须先转换才能做算术
print(f"user_typed + '1'      = {user_typed + '1'}")        # 字符串相加是拼接 → '181'
print("↑ 注意最后一行：'18' + '1' 得到 '181'，不是 19。这个 bug 极其隐蔽。")
print("  下面这行故意注释掉了，取消注释看看报什么错：")
print(user_typed + 1)              # TypeError: 不能把 str 和 int 相加


print()
# ── bool() 的转换规则：这些值都算 False ──
falsy_values = [0, 0.0, "", [], {}, None, False]
for v in falsy_values:
    print(f"bool({v!r}) = {bool(v)}")
print("↑ 上面这些叫「假值」。除此之外几乎都是 True，比如 bool(-1)、bool('0') 都是 True")
print(f"注意 bool('0') = {bool('0')}  ← 非空字符串就是 True，哪怕内容是 0")
````

## 3. 字符串操作

### 3.1 f-string：最重要的技能，没有之一

在引号前面加一个 `f`，就能在字符串里用 `{}` 直接插入变量或表达式。
这是现代 Python 的标准写法，你以后 90% 的字符串拼接都用它。

````python
name = "青承"
day = 2
progress = 2 / 84

# ── 基本用法 ──
print(f"你好，{name}")
print(f"今天是第 {day} 天")

# ── {} 里可以放任意表达式，不只是变量 ──
print(f"明天是第 {day + 1} 天")
print(f"{day} 的平方是 {day ** 2}")

# ── 格式化数字：{值:格式} ──
print(f"进度（原始）  : {progress}")
print(f"进度（2位小数）: {progress:.2f}")      # .2f = 保留2位小数
print(f"进度（百分比） : {progress:.1%}")       # .1% = 百分比保留1位
print(f"补零对齐      : {day:03d}")            # 03d = 补齐3位，前面填0

# ── 对齐与宽度，做表格时很有用 ──
for item, count in [("苹果", 3), ("香蕉", 12), ("樱桃", 108)]:
    print(f"{item:<6}{count:>5}")              # <6 左对齐占6格，>5 右对齐占5格

# ── 在 f-string 里显示引号和大括号本身 ──
print(f"大括号要写成双份: {{这样}}")
print(f"!r 会带上引号: {name!r}")              # 调试时很有用，能看出有没有多余空格
````

### 3.2 切片：从字符串里截一段

语法是 `s[开始:结束:步长]`，**含头不含尾**（包含开始位置，不包含结束位置）。
编号从 0 开始，负数表示从后往前数。

````python
s = "Python编程"

print(f"原字符串      : {s}")
print(f"s[0]         : {s[0]}")        # 第 0 个字符
print(f"s[-1]        : {s[-1]}")       # 最后一个字符
print(f"s[0:3]       : {s[0:3]}")      # 第0到第2个（不含第3个）
print(f"s[:3]        : {s[:3]}")       # 省略开始 = 从头开始
print(f"s[3:]        : {s[3:]}")       # 省略结束 = 一直到末尾
print(f"s[:]         : {s[:]}")        # 全都省略 = 完整复制一份
print(f"s[::2]       : {s[::2]}")      # 步长2，每隔一个取一个
print(f"s[::-1]      : {s[::-1]}")     # 步长-1 = 反转字符串（经典技巧）

print()
print("── 切片永远不会报错，越界了就是取到能取的为止 ──")
print(f"s[0:100]     : {s[0:100]}")   # 只有 8 个字符，但不报错
print(f"s[50:60]     : '{s[50:60]}'")  # 空字符串，也不报错

print()
print("── 对比：直接用下标越界就会报错 ──")
try:
    print(s[100])
except IndexError as e:
    print(f"IndexError: {e}")
````

### 3.3 最常用的字符串方法

这些是**方法**（用圆括号调用），不是函数。写法是 `字符串.方法名(参数)`。

````python
raw = "  报关单号 , TKX-2026-001 , 越南  "

print(f"原始          : {raw!r}")          # !r 能显示出空格，调试必备
print(f"strip()       : {raw.strip()!r}")  # 去掉首尾空白（不去中间的）
print(f"lstrip/rstrip : {raw.lstrip()!r}")

print()
cleaned = raw.strip()
parts = cleaned.split(",")                  # 按逗号切成列表
print(f"split(',')    : {parts}")
print(f"切出来是列表，有 {len(parts)} 个元素")

print()
# 切出来的每段还带着空格，要逐个 strip
tidy = [p.strip() for p in parts]           # 这是列表推导式，第1周会正式讲
print(f"逐段清理后     : {tidy}")
print(f"取第 2 段      : {tidy[1]!r}")      # ← 方括号按位置取值，昨天学的

print()
print("── 其他常用方法 ──")
text = "AI Product Journey"
print(f"replace()  : {text.replace('AI', '人工智能')}")   # 替换
print(f"upper()    : {text.upper()}")
print(f"lower()    : {text.lower()}")
print(f"find()     : {text.find('Product')}")            # 找到返回位置，找不到返回 -1
print(f"count()    : {text.count('n')}")                 # 出现几次
print(f"startswith : {text.startswith('AI')}")           # 返回 bool
print(f"endswith   : {text.endswith('.py')}")
print(f"isdigit()  : {'123'.isdigit()} / {'12a'.isdigit()}")

print()
print("── ⚠️ 重要：字符串是不可变的，方法不会改变原字符串 ──")
print(f"text 还是     : {text!r}")
print("所有方法都是返回一个新字符串。要保留结果必须赋值：text = text.upper()")
````

## 4. 运算符：重点是 `//` 和 `%`

| 运算符 | 名称 | 例子 | 结果 |
|---|---|---|---|
| `+` `-` `*` | 加减乘 | `7 * 3` | `21` |
| `/` | 除法（**永远返回 float**） | `7 / 2` | `3.5` |
| `//` | 整除（向下取整） | `7 // 2` | `3` |
| `%` | 取余数 | `7 % 2` | `1` |
| `**` | 幂 | `7 ** 2` | `49` |

`//` 和 `%` 是一对搭档：**`a // b` 得到「有几份」，`a % b` 得到「还剩几个」**。
这个组合在处理时间、分页、批次时天天用。

````python
print("── / 和 // 的区别 ──")
print(f"7 / 2   = {7 / 2}")       # 3.5  永远是浮点数
print(f"6 / 2   = {6 / 2}")       # 3.0  即使整除也是浮点数！
print(f"7 // 2  = {7 // 2}")      # 3    整数
print(f"6 // 2  = {6 // 2}")      # 3    整数

print()
print("── ⚠️ 负数的 // 是「向下」取整，不是「向零」 ──")
print(f"-7 // 2  = {-7 // 2}")     # -4，不是 -3！向下 = 往更小的方向
print(f"-7 / 2   = {-7 / 2}")      # -3.5
print(f"7 // -2  = {7 // -2}")     # -4

print()
print("── // 和 % 搭档的经典用法：时间换算 ──")
total_seconds = 3725
hours = total_seconds // 3600              # 有几个完整的小时
remainder = total_seconds % 3600           # 去掉小时后还剩多少秒
minutes = remainder // 60                  # 剩下的秒里有几个完整的分钟
seconds = remainder % 60                   # 最后剩下的秒
print(f"{total_seconds} 秒 = {hours} 小时 {minutes} 分 {seconds} 秒")

print()
print("── 另一个经典用法：判断奇偶 ──")
for n in range(1, 7):
    label = "偶数" if n % 2 == 0 else "奇数"
    print(f"  {n} 是{label}")

print()
print("── 分页计算（第 3 周处理数据时会用到）──")
total_rows, page_size = 95, 10
total_pages = (total_rows + page_size - 1) // page_size    # 向上取整的技巧
print(f"{total_rows} 条数据，每页 {page_size} 条，共 {total_pages} 页")
print(f"对比昨天的写法：(95 - 1) // 10 + 1 = {(95 - 1) // 10 + 1}  ← 两种等价")
````
