## 1. `for` 循环与 `range()`　*约 8 分钟*

循环 = 把一段代码重复执行多次。Python 最常用的循环是 `for`。

```
for 临时变量 in 一串东西:
    对每个东西做的事        ← 缩进的部分会被重复执行
```

`range()` 是最常配的「造一串数字」的工具，三种写法：

| 写法 | 含义 | 产生的值 |
|---|---|---|
| `range(5)` | 从 0 开始，5 个 | 0 1 2 3 4 |
| `range(1, 6)` | 从 1 到 5（**不含 6**） | 1 2 3 4 5 |
| `range(0, 10, 2)` | 从 0 到 9，步长 2 | 0 2 4 6 8 |

注意：**含头不含尾**——和昨天学的切片规则一模一样。

````python
print("range(5)      →", list(range(5)))
print("range(1, 6)   →", list(range(1, 6)))
print("range(0, 10, 2) →", list(range(0, 10, 2)))
print("range(3, 3)   →", list(range(3, 3)), " ← 空！起点等于终点，一个都没有")
print()
print("=== for 循环实际跑起来 ===")
for n in range(1, 6):
    print(f"  第 {n} 次循环，n 现在的值是 {n}")
print("循环结束，n 停在最后一个值:", n)
````

````python
# 累加是循环最常见的用途
total = 0
for n in range(1, 101):     # 1 到 100
    total = total + n       # 也可以简写成 total += n
print(f"1 加到 100 = {total}")

# 昨天的复习题就是这个结构：循环里套 if
total_even = 0
for n in [1, 2, 3, 4, 5]:
    if n % 2 == 0:          # n 是偶数
        total_even = total_even + n
print(f"其中偶数的和 = {total_even}")   # 2 + 4 = 6
````

## 2. `while` 循环与死循环　*约 6 分钟*

`for` 是「知道要跑几次」（有一串东西可遍历）。
`while` 是「**不知道要跑几次，但知道什么时候该停**」——只要条件成立就一直跑。

```
while 条件:
    做的事
```

**⚠️ 致命风险**：如果你忘了在循环体里改变那个条件，它就永远成立，程序永远不结束——这叫**死循环**。新手最容易在这里失控（CPU 狂转，只能强制关掉）。

````python
# 正确的 while：每轮都让计数器 +1，条件迟早变成 False
count = 0
while count < 5:
    print(f"  count = {count}")
    count = count + 1          # ← 这行是关键！没有它就会永远打印 0
print("循环正常结束，count 最终 =", count)
````

````python
# while count < 5:
#     print("我要永远跑下去了")   # ← count 从来没被改过，条件永远 True
#
# 这就是死循环。症状：程序卡住不返回、风扇狂转。
# 怎么脱困：按 Ctrl + C 强制中断（终端里），或者关掉内核（notebook 里点停止按钮）。
#
# 你前天在终端遇到 dquote> 卡住时用的就是同一个思路：Ctrl+C 是万能自救键。
print("记住这条：写 while 时，第一件事是想清楚『哪个变量在推动我走向退出条件』。")
print("如果想不出这个变量，说明这题该用 for 而不是 while。")
````

````python
# while 的真实用途：跑到达标为止（不知道要跑几轮）
# 例子：复利，本金多少年能翻倍
money = 10000
years = 0
while money < 20000:
    money = money * 1.05        # 每年 5% 复利
    years = years + 1
print(f"本金 10000，年息 5%，{years} 年翻倍（最终 {money:,.0f} 元）")
# 这就是 for 做不到的：你事先不知道要循环多少次
````

## 3. `break` 与 `continue`　*约 5 分钟*

- `break` —— **整个循环立刻结束**，跳出循环体
- `continue` —— **跳过本轮**剩下的代码，直接进入下一轮

一个比喻：break 是「不干了，走人」，continue 是「这个跳过，下一个」。

````python
print("=== break：找到目标就停，后面的不再看 ===")
nums = [3, 7, 2, 9, 5]
for n in nums:
    if n == 2:
        print(f"  找到了 {n}，停止查找")
        break                   # 后面的 9、5 根本不会被遍历到
    print(f"  检查 {n}，不是目标")

print()
print("=== continue：跳过这一轮，其他照常 ===")
for n in range(1, 8):
    if n % 2 == 0:
        continue               # 偶数就跳过，下面的 print 不执行
    print(f"  奇数: {n}")
````

**🔮 预测**：下面这段输出什么？特别注意 `break` 之后程序走到哪。

````python
print("第一段:")
for i in range(1, 5):
    if i == 3:
        break
    print(i)
print("第二段:")
for i in range(1, 5):
    if i == 3:
        continue
    print(i)
print("循环全部结束")
# i = 1
# i = 2
# 第二段:
# 1
# 2
# 4
# 循环全部结束 
````

> ### ⏸ 检查点 3
> `break` 完全退出循环；`continue` 只跳过本轮。这两个搞混是常见的 bug 源头。

## 4. 遍历各种容器　*约 6 分钟*

循环真正的用法是「遍历数据」。你已经学过列表和字典，for 对它们的遍历方式不同——**这块最容易写出 bug**。

````python
fruits = ["苹果", "香蕉", "橙子"]
person = {"name": "张三", "age": 30, "city": "上海"}

print("=== 遍历列表：直接拿到每个元素 ===")
for f in fruits:
    print("  ", f)

print()
print("=== 遍历字典的【键】—— 注意！默认只给键 ===")
for k in person:                 # 不加 .values() 就是遍历键
    print(f"   键={k}  值={person[k]}")     # 想要值必须用 person[k] 再取一次

print()
print("=== 遍历字典更清楚的写法：items() ===")
for key, value in person.items():        # 一次拿到键和值
    print(f"   {key} → {value}")

print()
print("=== 遍历字符串：一个字符一个字符 ===")
for ch in "AI产品":
    print("   字符:", repr(ch))
````

**⚠️ 一个必须记住的坑**：直接 `for x in 字典` 拿到的是**键**，不是值，也不是「键值对」。
这是从其他语言过来的人最常踩的。想要值就用 `d.items()` 或 `d[key]`。

````python
prices = {"苹果": 8, "香蕉": 3, "橙子": 5}

# 错误直觉：以为 for x in prices 能拿到价格 
print("遍历到的其实是键:", [x for x in prices])

# 想算总价，正确写法：
total = 0
for name, price in prices.items():
    total = total + price
print(f"总价 = {total}")

# 或者用昨天的 in 知识：
print("'苹果' in prices →", "苹果" in prices, "  ← in 查的是键")
print("'苹果' in prices.values() →", "苹果" in prices.values(), "  ← 值里查不到")
````

## 5. `enumerate()` 与 `zip()`　*约 7 分钟 —— 全 notebook 最有价值的一节*

这两个函数你以后**每天都会写**。它们解决的问题是：遍历列表时我还想要序号 / 我有两个列表想配着遍历。

### `enumerate()`：边遍历边拿编号

老写法（丑且容易错）：

````python
fruits = ["苹果", "香蕉", "橙子"]

# ❌ 老手写法，新手容易写错
index = 0
for f in fruits:
    print(f"  {index}: {f}")
    index = index + 1          # 忘了这行就死循环/逻辑错

print()
# ✅ 用 enumerate，一行搞定
for i, f in enumerate(fruits):
    print(f"  {i}: {f}")

print()
# 编号不从 0 开始？给第二个参数
for i, f in enumerate(fruits, start=1):
    print(f"  第{i}名: {f}")
````

### `zip()`：两个（或多个）列表配对

````python
names = ["张三", "李四", "王五"]
scores = [88, 95, 72]

print("=== zip 把两个列表捆成一对一对 ===")
for name, score in zip(names, scores):
    print(f"  {name} 的分数是 {score}")

print()
print("=== 长度不等时，zip 以短的为准（不会报错，直接截断）===")
a = [1, 2, 3, 4]
b = ["甲", "乙", "丙"]          # 少一个
print("   zip(a, b) →", list(zip(a, b)))
print("   ⚠️ 那个 4 被静悄悄丢掉了！这是很难发现的 bug")

print()
print("=== zip 常用于造字典（键列表 + 值列表）===")
print("   dict(zip(names, scores)) →", dict(zip(names, scores)))
````

## 6. 列表推导式：读懂即可　*约 7 分钟*

你已经见过它三次了（Day 1 进度条、Day 2 清洗、Day 4 上面）。今天正式讲。

```
[ 表达式 for 变量 in 容器 if 条件 ]
```

它是「用一行代码，从一个容器生成一个新列表」。从右往左读：
- `for 变量 in 容器` —— 遍历谁
- `if 条件` —— 只保留满足条件的（可省略）
- 最左边的 `表达式` —— 每个要变成什么

**这是 AI 生成的代码里最常见的写法**，你不一定要会写，但**必须能读懂**。

````python
nums = [1, 2, 3, 4, 5, 6]

# 基本形态：每个都平方
squares = [n ** 2 for n in nums]
print("平方:", squares)

# 带条件：只留偶数
evens = [n for n in nums if n % 2 == 0]
print("偶数:", evens)

# 表达式 + 条件：偶数的平方
even_squares = [n ** 2 for n in nums if n % 2 == 0]
print("偶数的平方:", even_squares)

print()
print("=== 它等价于这个普通循环（读懂推导式的关键）===")
result = []
for n in nums:
    if n % 2 == 0:
        result.append(n ** 2)
print("等价的循环写法:", result, "| 和推导式结果一样:", result == even_squares)

print()
print("=== 结合昨天的字符串：Day 2 你写的清洗就是这个 ===")
raw = ["  苹果 ", "香蕉  ", "  橙子"]
clean = [f.strip() for f in raw]
print("清洗前:", raw)
print("清洗后:", clean)
````

**⚠️ 推导式的两个限制**（新手容易过度使用）：

1. **一行套太复杂就别用**。嵌套两层以上的推导式几乎没法维护。
2. **推导式是「造新列表」的**，不是「做某件事」的。如果你只是想打印、只是想在原列表上操作，用普通 for，别用推导式。

````python
# 反例：这样用推导式很烂（虽然能跑）
# [print(x) for x in range(3)]     ← 目的是打印，却造了个没人用的列表，该用 for

# 正例：嵌套推导式能读但很难维护，别写
matrix = [[1, 2], [3, 4], [5, 6]]
flat_hard = [n for row in matrix for n in row]     # 两个 for，从右往左读
print("嵌套推导式结果:", flat_hard)

# 同样的事，普通循环谁都看得懂
flat_clear = []
for row in matrix:
    for n in row:
        flat_clear.append(n)
print("普通循环结果  :", flat_clear, "| 一致:", flat_hard == flat_clear)
print()
print("规则：两层以上，或者别人看不懂，就老老实实写 for 循环。可读性 > 短。")
````

## 7. 嵌套循环　*约 5 分钟 —— 选做，时间不够可跳*

循环里再套循环，用来处理「二维」的东西（表格、矩阵、乘法表）。

````python
print("=== 九九乘法表（最经典的嵌套练习）===")
for i in range(1, 10):
    row = ""
    for j in range(1, i + 1):
        row = row + f"{j}x{i}={i*j:<3}"     # :<3 左对齐占3格，让输出整齐
    print(row)

print()
print("外层跑 9 次，内层平均跑 4-5 次，总共约 45 次。")
print("⚠️ 嵌套循环的开销是「相乘」的：外层 1 万次 × 内层 1 万次 = 1 亿次。")
print("   第 3 周处理数据时，能避免嵌套就避免，这是性能的头号杀手。")
````

## 练习 A：FizzBuzz　*必做，约 5 分钟*

## 练习 B：词频统计　*必做，约 8 分钟*

**统计一段文字里每个词出现几次。** 这是你未来处理任何文本数据的基本功，
也是大模型应用里最常见的第一步（分词 + 计数）。

要求：
1. 用 `.split(" ")` 把句子切成一个个词
2. 用**字典**记录每个词出现了几次（键=词，值=次数）
3. 打印每个词和它的次数

**提示**：字典累加的经典写法，今天新学的一个惯用法：
```python
counts[word] = counts.get(word, 0) + 1
```
`.get(键, 默认值)` 的意思是：键存在就返回它的值，不存在就返回默认值（这里给 0）。
这一行省掉了「先判断键在不在」的麻烦。

````python
text = "产品 技术 产品 用户 技术 产品 数据 用户 产品"

counts = {}          # 空字典，用来装每个词的次数
for word in text.split():
    counts[word] = counts.get(word, 0) + 1
for word, n in counts.items():
    print(f"{word} 出现了 {n} 次")
# TODO: 遍历切分出来的词，用上面那行 counts.get 的写法累加
# 然后用 for word, n in counts.items() 打印，格式：'产品 出现了 4 次'

# 进阶（可选）：按次数从多到少排序打印
# 提示：sorted(counts.items(), key=lambda kv: kv[1], reverse=True)
#      这行现在看不懂没关系，Day 8 学函数后会明白
````
