## 1. list 的增删改查　*约 7 分钟*

列表用方括号 `[]` 创建，是一个**有序、可修改**的容器。

| 操作 | 写法 | 说明 |
|---|---|---|
| 查 | `lst[0]` `lst[-1]` `lst[1:3]` | 按位置，支持切片 |
| 增 | `lst.append(x)` | 尾部加一个（最常用） |
| 增 | `lst.insert(0, x)` | 指定位置插入 |
| 改 | `lst[0] = x` | 按位置赋值 |
| 删 | `lst.remove(x)` | 按**值**删第一个匹配 |
| 删 | `x = lst.pop()` | 弹出最后一个并返回它 |
| 删 | `del lst[0]` | 按位置删 |

````python
fruits = ["苹果", "香蕉"]

# ── 增 ──
fruits.append("橙子")                 # 尾部加，最常用
fruits.insert(0, "榴莲")              # 指定位置插
print("append/insert 后:", fruits)   #append/insert 后: ['榴莲', '苹果', '香蕉', '橙子']

# ── 查 ──
print("第一个:", fruits[0]) # 第一个: 榴莲
print("最后一个:", fruits[-1])  # 最后一个: 橙子
print("切片 [1:3]:", fruits[1:3])  # 切片 [1:3]: ['苹果', '香蕉']
print("长度:", len(fruits), "| 香蕉在哪:", fruits.index("香蕉")) # 长度: 4 | 香蕉在哪: 2
print("在不在里面:", "苹果" in fruits, "|", "葡萄" in fruits) # 在不在里面: True | False

# ── 改 ──
fruits[0] = "红苹果"
print("改完之后:", fruits) # 改完之后: ['红苹果', '苹果', '香蕉', '橙子']
# ── 删 ──
fruits.remove("红苹果")               # 按值删
print("remove 后:", fruits)          # remove 后: ['苹果', '香蕉', '橙子']
last = fruits.pop()                   # 弹出最后一个，并拿到它的值
print("pop 拿到的值:", last, "| 剩余:", fruits)  # pop 拿到的值: 橙子 | 剩余: ['苹果', '香蕉']

# ── 计数（这个和第 3 节的字典累加是同一类需求）──
data = ["A", "B", "A", "C", "A"]
print()
print(f"'A' 出现了 {data.count('A')} 次") #A 出现了 3 次
````

## 2. 排序：`sort()` 还是 `sorted()`　*约 5 分钟*

这两个东西长得很像，行为完全不同，**是最容易搞混的一对**。

````python
nums = [3, 1, 4, 1, 5]

# sort()  → 原地排序，改的是自己，返回 None
result_of_sort = nums.sort()
print("nums.sort() 之后 nums =", nums)
print("  但 sort() 的返回值是:", result_of_sort, " ← None！")

# 所以千万不要这么写：
lst = [3, 1, 2]
broken = lst.sort()
print("  写成 broken = lst.sort() → broken 是", broken, "（这就是新手常犯的错）")

print()
# sorted() → 不动原列表，返回一个新的排好序的列表
original = [3, 1, 4, 1, 5]
new_list = sorted(original)
print("sorted() 返回新列表:", new_list)
print("原列表【没被改动】  :", original)
print()
print("倒序: sorted(original, reverse=True) →", sorted(original, reverse=True))
````

### 按某个字段排序（处理真实数据必备）

你的数据往往是「列表套字典」，要按某个键排：

````python
orders = [
    {"品名": "光伏玻璃", "金额": 125000},
    {"品名": "胶膜", "金额": 89000},
    {"品名": "背板", "金额": 234000},
]

# 按金额从大到小排
by_amount = sorted(orders, key=lambda o: o["金额"], reverse=True)
for o in by_amount:
    print(f'  {o["品名"]:<8} {o["金额"]:>8,}')

print()
print("现在不用完全看懂 lambda，只要记住这个形状：")
print("  sorted(列表, key=lambda 元素: 元素[按哪个键], reverse=True)")
print("Day 8 学函数时会彻底讲清楚。今天会套这个模板就行。")
````

## 3. ⚠️ 赋值不是复制　*约 8 分钟 —— 今天最重要的一节*

**这一节如果你只学一个东西，就学它。** 它造成的 bug 数量，超过前面所有语法错误的总和。

看这个：

````python
a = [1, 2, 3]
b = a                 # 你以为这是「把 a 复制一份给 b」
b.append(99)          # 我改 b，关 a 什么事？

print("b =", b)
print("a =", a, " ← a 也跟着变了！！")
print()
print("因为 b = a 复制的是【引用】（那个列表的地址），不是列表本身。")
print("a 和 b 是同一个列表的两个名字，就像一个人的大名和小名。")
````

````python
# 对比：数字就没这个问题
x = 10
y = x
y = 20
print(f"x = {x}, y = {y}   ← 数字是各自独立的")
print()

# 列表却是共享的
p = [1, 2]
q = p
q[0] = 999
print(f"p = {p}, q = {q}   ← 两个都变了")
print()
print("=== 用昨天学的 is 验证它俩是不是同一个东西 ===")
print(f"p is q  → {p is q}    同一个对象")
print(f"p == q  → {p == q}    内容也相同")
````

### 想要真正的「独立复制」，三种正确写法

````python
original = [1, 2, 3]

copy1 = original[:]              # 切片复制（你 Day 2 学的 s[:] 就是这个用途）
copy2 = list(original)           # 用 list() 造一个新的
copy3 = original.copy()          # 显式 .copy() 方法

copy1.append(99)
copy2.append(88)
copy3.append(77)
print("原列表:", original, " ← 安然无恙")
print("copy1:", copy1)
print("copy2:", copy2)
print("copy3:", copy3)
print()
for name, c in (("copy1", copy1), ("copy2", copy2), ("copy3", copy3)):
    print(f"  {name} is original → {c is original}")
````

**⚠️ 但注意：上面三种都只防住第一层。** 列表里套了字典时，内层还是共享的：

````python
deep = [{"品名": "玻璃", "数量": 3}]
shallow = deep.copy()             # 浅拷贝

# 改外层（增删列表元素）→ 互不影响
shallow.append({"品名": "胶片"})
print("改外层之后，原列表长度:", len(deep), "  副本长度:", len(shallow))

# 改内层（动那个字典）→ 【两个一起变】！
shallow[0]["数量"] = 999
print("改内层之后:", deep)
print("  ↑ 原列表里的数量也变成 999 了！因为 copy() 只复制了外层壳子")
print()
print("想彻底断开，用 deepcopy：")
import copy
real_deep = [{"数量": 3}]
safe = copy.deepcopy(real_deep)
safe[0]["数量"] = 7
print("  deepcopy 之后原数据:", real_deep, " 副本:", safe, " ← 互不干扰")
print()
print("现阶段你只需要记住两件事：")
print("  1) b = a 对列表/字典是共享，不是复制")
print("  2) 遇到数据莫名其妙自己变了，第一个怀疑对象就是这里")
````

## 4. dict：键值操作与 `get()`　*约 7 分钟*

字典用花括号 `{}` 创建，存的是「键 → 值」的对应关系。**这是 Python 里最重要的数据结构。**

| 操作 | 写法 | 注意 |
|---|---|---|
| 查 | `d["键"]` | 键不存在会 **KeyError** |
| 查（安全） | `d.get("键", 默认值)` | 不存在返回默认值，不报错 |
| 增/改 | `d["键"] = 值` | 有就改，没有就加，同一句 |
| 删 | `del d["键"]` | |
| 判断 | `"键" in d` | **查的是键，不是值** |
| 全部键 | `d.keys()` | |
| 全部值 | `d.values()` | |
| 全部对 | `d.items()` | |

````python
order = {"单号": "TKX-001", "国家": "越南", "金额": 125000}

# 查
print("金额:", order["金额"])
print("所有键:", list(order.keys()))
print("所有值:", list(order.values()))

# ⚠️ 键不存在：两种方式的分水岭
print()
try:
    print(order["汇率"])
except KeyError as e:
    print(f"直接取值 KeyError: 键 {e} 不存在 → 程序会当场崩")
print(f"用 get(): {order.get('汇率')}   ← 返回 None，不崩")
print(f"get() 给默认值: {order.get('汇率', 0)}")
print()
print("  规则：不确定键存在吗 → 一律用 get()。这是防御性编程的第一课。")

# 增和改是同一句
order["状态"] = "已放行"        # 新键 → 增
order["状态"] = "已结关"        # 已有键 → 改
print("增改之后 状态:", order["状态"])

# 删除
del order["国家"]
print("删掉国家后的键:", list(order.keys()))
````

**昨天你已经用过 `get()` 了**——词频统计那行 `counts[word] = counts.get(word, 0) + 1`。
现在正式理解它：键不存在时 `get` 返回 0，加 1 变成 1；键已存在时返回旧次数，加 1 递增。
**这一行是 Python 最经典的惯用法之一，你会反复遇到。**

## 5. 遍历字典（昨天那个坑的完整版）　*约 5 分钟*

````python
person = {"姓名": "张三", "年龄": 30, "城市": "上海"}

print("① 直接遍历 → 拿到的是【键】")
for x in person:
    print(f"   {x!r}")

print()
print("② 遍历键，再用键取值")
for k in person:
    print(f"   {k} = {person[k]}")

print()
print("③ 推荐写法：items() 一次拿到键和值")
for k, v in person.items():
    print(f"   {k} = {v}")

print()
print("④ 只要值：values()")
for v in person.values():
    print(f"   {v}")

print()
print("=== 为什么默认遍历键？=== ")
print("因为字典的核心用途是【用键找值】，遍历键是最常见需求，所以设成默认。")
print("记住这条就不会错：for 循环遍历字典，拿到的是键。")
````

## 6. JSON 与嵌套结构　*约 10 分钟 —— 今天的核心产出*

**这是你今天必须练熟的东西。** 真实世界的接口返回、配置文件、数据库导出，几乎都是这个形状：

```
列表 里面装着 字典，字典的值里又装着 列表/字典
```

JSON（一种文本格式）解析进 Python 之后，**就是字典和列表**。学会在嵌套结构里-navigation 是今天的目标。

先看一个「列表套字典」：

````python
# 这是一份报关单的货物明细
declaration = {
    "单号": "TKX-2026-001",
    "申报日期": "2026-10-20",
    "国家": "越南",
    "货物": [
        {"品名": "光伏玻璃", "数量": 1200, "单位": "平方米", "单价": 85.5},
        {"品名": "胶膜", "数量": 300, "单位": "卷", "单价": 1200.0},
        {"品名": "铝边框", "数量": 8000, "单位": "根", "单价": 12.8},
    ],
    "备注": {"客户": "Flat Vietnam", "加急": True},
}

print("这是一份【字典】，键有:", list(declaration.keys()))
print()
print("=== 逐层往里剥 ===")
print(f'  declaration["国家"]            → {declaration["国家"]}')
print(f'  declaration["货物"]            → 是个【{type(declaration["货物"]).__name__}】，{len(declaration["货物"])} 项')
print(f'  declaration["货物"][0]         → {declaration["货物"][0]["品名"]} （第一个货物，它是个字典）')
print(f'  declaration["货物"][0]["数量"]  → {declaration["货物"][0]["数量"]}')
print(f'  declaration["备注"]["客户"]     → {declaration["备注"]["客户"]}')
print()
print("读法口诀：从左往右，【一层一层剥】。方括号每剥一层，先想清楚剥出来的是什么类型。")
````

````python
# 实战：算总金额 = Σ(数量 × 单价)
total = 0
print("逐项计算:")
for item in declaration["货物"]:
    subtotal = item["数量"] * item["单价"]
    total += subtotal
    print(f'  {item["品名"]:<8} {item["数量"]:>6} × {item["单价"]:>8} = {subtotal:>12,.2f}')

print(f"\n  合计: {total:,.2f}")
print()
print("这个 for + 字典取键 + 累加 的组合，就是数据处理的最小骨架。")
print("第 3 周学 Pandas 之后，同样一件事会变成一个函数调用，但你得先懂底层。")
````

### 文本形式的 JSON 怎么变字典

`json` 模块负责「文本 ↔ 字典」的转换。API 返回给你的永远是**文本**，必须解析才能用。

````python
import json

# 这是【字符串】——API 返回的就是这个形状，注意它是引号包着的
raw_text = '{"单号": "TKX-002", "金额": 88000, "已放行": true, "明细": [{"品名": "玻璃"}]}'
print(f"raw_text 的类型: {type(raw_text).__name__} ← 是文本，不是字典！")
print('  所以 raw_text["单号"] 会报错 ↓')
try:
    raw_text["单号"]
except TypeError as e:
    print(f"  TypeError: {e}")

print()
data = json.loads(raw_text)              # 文本 → 字典（loads = load string）
print(f"json.loads 之后类型: {type(data).__name__}")
print(f"现在可以正常取值: data['单号'] = {data['单号']}")
print(f"data['金额']   = {data['金额']}")
print(f"data['已放行']  = {data['已放行']}    ← JSON 的 true 变成了 Python 的 True")
print(f"data['明细'][0]['品名'] = {data['明细'][0]['品名']}")

print()
print("=== 反向：字典 → 文本（存文件、发请求时用）===")
text_again = json.dumps(data, ensure_ascii=False, indent=2)
print(text_again[:180])
print("  ↑ ensure_ascii=False 才能正常显示中文，默认会变成 \\uXXXX 乱码")
````

**记这四个函数，第 8 周调大模型 API 天天用**：

| 函数 | 方向 | 场景 |
|---|---|---|
| `json.loads(文本)` | 文本 → 字典 | 收到 API 响应后解析 |
| `json.dumps(字典)` | 字典 → 文本 | 发请求前序列化 |
| `json.load(文件对象)` | 读文件 → 字典 | 读 .json 文件（少个 s） |
| `json.dump(字典, 文件对象)` | 字典 → 写文件 | 存成 .json 文件 |

**顺带说清一个你迟早会困惑的点**：JSON 用 `true/false/null`，Python 用 `True/False/None`。
`json.loads` 会自动翻译；但你手写 Python 代码时得用 Python 那套。

## 7. tuple 与 set　*约 6 分钟 —— 了解即可，不用练熟*

这两个用得少，但读别人代码时会碰到。

````python
# ── tuple 元组：不能改的列表 ──
point = (120.5, 31.2)          # 圆括号
print(f"tuple: {point}  类型: {type(point).__name__}")
print(f"  可以取: point[0] = {point[0]}")
try:
    point[0] = 999
except TypeError as e:
    print(f"  但不能改: TypeError: {e}")
print()
print("  什么时候用：经纬度、(年月日)、函数返回多个值。这些「不该被改」的一组数据。")
print(f"  你其实一直在用：for k, v in d.items() 里的 k, v 就是元组解包")
print(f"  sorted 那个 key=lambda 也是元组场景")

# ── set 集合：自动去重、只关心「有没有」 ──
print()
tags = ["光伏", "玻璃", "光伏", "越南", "玻璃"]
unique = set(tags)
print(f"list → set 自动去重: {tags}")
print(f"                     → {unique}")
print(f"  集合擅长判断存在（比列表快很多）: '光伏' in unique → {'光伏' in unique}")
print()
a = {"光伏", "玻璃", "越南"}
b = {"玻璃", "越南", "泰国"}
print(f"  交集 a & b = {a & b}")
print(f"  并集 a | b = {a | b}")
print(f"  差集 a - b = {a - b}   ← 在a但不在b的")
````

> ### ⏸ 检查点 6
> tuple 和 set 现在**不需要练熟**。记住两句话就够：
> 「tuple 是不能改的列表」，「set 是自动去重 + 快速判断有没有」。

## 练习 A：解析订单 JSON，算总金额　*必做，约 8 分钟*

下面这段是「从 API 拿到的原始文本」（注意它是字符串！）。要求：

1. 把它解析成能用的数据（提示：`json.loads`）
2. 打印订单号和客户名
3. **算出订单总额** = Σ(每行项目的 `数量 × 单价`)
4. 找出金额最大的那一行的品名
5. 打印时金额带千位分隔符和两位小数

**自测标准**（这三个值对上就说明你做对了）：

- 总额 = `370,643.60`
- 金额最大的那一行 = `胶膜`（240,000）
- 客户名称 = `Flat Vietnam`（要剥两层才取得到）

对不上就回头检查你的循环——**但先别改这三个标准值**，它们我用暴力法算过两遍。

````python
raw = '{\n  "订单号": "ORD-2026-1023",\n  "客户": {"名称": "Flat Vietnam", "国家": "VN"},\n  "状态": "confirmed",\n  "项目": [\n    {"品名": "光伏玻璃", "数量": 1500, "单价": 85.5},\n    {"品名": "胶膜", "数量": 200, "单价": 1200.0},\n    {"品名": "铝边框", "数量": 187, "单价": 12.8}\n  ]\n}'

# TODO 从这里开始写
data = json.loads(raw)      # 第1步: data = json.loads(raw)      打印 type(data) 确认它现在是字典
print(type(data))
print("订单号:", data["订单号"])
print("客户名称:", data["客户"]["名称"])
total = 0
for item in data["项目"]:# 第3步: 遍历 data["项目"] 累加 数量*单价
    amount = item["数量"] * item["单价"]
    total += amount
print("总金额:", total)
max_item = max(data["项目"], key=lambda p: p["数量"] * p["单价"])  # 第4步: 找最大金额那行（提示：max(data["项目"], key=lambda p: p["数量"]*p["单价"])）
print("最大金额的项目:", max_item["品名"], f"{max_item['数量'] * max_item['单价']:,.0f}")   #        或者先不用 max，用循环里比较大小也行，金额千分号
````

## 练习 B：按类别汇总　*必做，约 7 分钟*

**这就是词频统计的进阶版**，也是第 3 周 Pandas 的 `groupby` 手工版。
把货物按「类别」分组，算每类总金额。核心还是那个 `counts.get(k, 0) + 增量` 的模式。

````python
goods = [
    {"品名": "超白玻璃", "类别": "玻璃", "金额": 128250},
    {"品名": "EVA胶膜", "类别": "胶膜", "金额": 240000},
    {"品名": "镀膜玻璃", "类别": "玻璃", "金额": 95000},
    {"品名": "铝边框", "类别": "边框", "金额": 240000},
    {"品名": "背板", "类别": "化工", "金额": 88000},
    {"品名": "光伏玻璃", "类别": "玻璃", "金额": 60000},
]

# TODO:
by_category = {}  # 1) 建一个空字典 by_category
for good in goods:
    category = good["类别"]  # 2) 遍历 goods，把同类别的【金额累加】进去
    by_category[category] = by_category.get(category, 0) + good["金额"] #    提示：by_category[类别] = by_category.get(类别, 0) + 该行金额
print(sorted(by_category.items(), key=lambda x: x[1], reverse=True))  # 3) 打印每个类别的总金额，按金额从大到小排
#
# 期望结果（4 个类别，注意「铝边框」这条的类别字段值是「边框」不是品名）：
#   玻璃 283,250  /  胶膜 240,000  /  边框 240,000  /  化工 88,000
# 做完想一想：这个「分组累加」在你工作里能用来干嘛？
````

## 练习 C：清洗 + 去重　*选做，约 6 分钟*

给你一份**脏数据**：有重复项、有前后空格、有大小写混用、还有空字符串。
要求输出去重、清洗干净的列表。

**这题的隐藏考点**：`set` 能去重但不保证顺序。如果你要「保留第一次出现的顺序」，
`set` 就不够用了——需要用「set 记录见过什么 + list 保持顺序」的组合。

````python
raw_list = ["  光伏玻璃", "胶膜", "光伏玻璃 ", "EVA", "eva", "铝边框", "", "  ", "胶膜", "背板"]
# TODO:
# 1) 去掉每个元素的首尾空格（列表推导式，你 Day 2 就写过这个）
# 2) 去掉空字符串（提示：if 空字符串是假值，直接 if x 就行）
# 3) 统一大小写（想想为什么 "EVA" 和 "eva" 会被当成两个不同的值）
# 4) 去重，并且【保留第一次出现的顺序】
#
# 期望最终结果（忽略大小写去重后）:
#   ['光伏玻璃', '胶膜', 'eva', '铝边框', '背板']
#
# 提示：用一个 seen = set() 记录已经见过的，遍历原列表，
#      没见过就加进结果列表并同时加进 seen。这是保序去重的标准写法。
seen = set()
strip_list = [x.strip() for x in raw_list]
if_list = [x for x in strip_list if x]
final_list = []
for x in if_list:
    lower_x = x.lower()
    if lower_x not in seen:
        final_list.append(x)
        seen.add(lower_x)
print(final_list)
````
