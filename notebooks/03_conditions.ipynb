## 1. 比较运算符：结果都是「真或假」

六个比较符号，它们的返回值永远是布尔值 `True` / `False`（昨天学过 bool）：

| 符号 | 意思 | 注意 |
|---|---|---|
| `==` | 等于 | **两个等号**！一个 `=` 是赋值 |
| `!=` | 不等于 | 没有 `=>` 这种写法 |
| `>` `<` `>=` `<=` | 大于/小于/大于等于/小于等于 | |

````python
print(5 > 3)      # True
print(5 >= 5)     # True   等号包含在内
print(5 == 5)     # True   比较用的是双等号
print(5 != 3)     # True
print(5 < 3)      # False

# 承接昨天的坑：数字和文本用 == 比较，永远是 False
print()
print(f"5 == '5'      → {5 == '5'}      ← 一个是数字一个是文本，不相等")
print(f"int('5') == 5 → {int('5') == 5}   ← 先把文本转成数字才相等")
print(f"type(5 > 3)   → {type(5 > 3)}    ← 比较的结果是布尔值")
````

## 2. `if` 语句与缩进

```
if 条件:
    条件成立时才做的事      ← 这一行必须有缩进
```

**缩进在 Python 里不是「为了好看」，是语法**。别人写的代码缩进错了会直接报错。标准是 **4 个空格**。

最容易犯的错：把 `if 温度 = 30` 写成赋值。`=` 是「装进去」，`==` 才是「比一比」。

````python
temperature = 35

if temperature > 30:
    print("太热了")       # 有缩进 → 属于 if，条件成立才执行
    print("记得开空调")    # 有缩进 → 也属于 if
print("这段永远执行")      # 没缩进 → 不属于 if，无论如何都会跑

print()
print("换个低温度看：")
temperature = 20
if temperature > 30:
    print("这句话不会出现")   # 条件不成立，整块被跳过
print("程序继续正常往下")
````

**⚠️ 两个新手必踩的语法错**（下面都注释掉了，取消注释感受一下报错）：

````python
x = 10

# 错误 1：用单等号做比较
if x == 10:          # SyntaxError: invalid syntax
     pass

# 错误 2：if 那行末尾漏了冒号
if x > 5:            # SyntaxError: expected ':'
     print("hi")

# 错误 3：该缩进的没缩进
if x > 5:
     print("hi")          # IndentationError: expected an indented block

print("上面三个如果都取消注释，会分别报这三种错。看到它们，你就知道问题出在哪。")
````

## 3. `if / elif / else`：多分支

只有「是或否」两个分支用 `if / else`。有多个可能时，用 `elif`（else if 的缩写）。

**关键规则：`elif` 只会执行「第一个成立的」那个分支，然后整个 if 结构就结束了。** 后面的不再判断。

````python
score = 85

if score >= 90:
    grade = "优秀"
elif score >= 80:          # 上一行不成立，才轮到这行
    grade = "良好"
elif score >= 60:
    grade = "及格"
else:                      # 前面全不成立才走这里
    grade = "不及格"

print(f"分数 {score} → {grade}")
print()
print("注意：85 同时满足 '>=80' 和 '>=60'，但因为 >=90 不成立、>=80 先成立，")
print("拿到『良好』后就跳出，不会再看 >=60。这就是「只执行第一个成立的」。")
````

**⚠️ 分支顺序很重要！** 当条件区间重叠时，写反了会全错。看这个反面例子：

````python
score = 95

# 错误写法：从低到高判断
if score >= 60:
    g1 = "及格"          # 95 也满足 >=60！第一个就撞上，后面永远轮不到
elif score >= 90:
    g1 = "优秀"
else:
    g1 = "不及格"

# 正确写法：从高到低判断
if score >= 90:
    g2 = "优秀"          # 先判最严的条件
elif score >= 60:
    g2 = "及格"
else:
    g2 = "不及格"

print(f"从低到高判(错)  : {g1}")     # 会得到『及格』
print(f"从高到低判(对)  : {g2}")     # 得到『优秀』
print()
print("口诀：区间重叠时，把最『苛刻』的条件放最前面。")
````

## 4. `and` `or` `not`：组合条件

| 符号 | 意思 | 结果 |
|---|---|---|
| `A and B` | A **且** B | 两个都 True 才 True |
| `A or B` | A **或** B | 只要有一个 True 就 True |
| `not A` | 非 A | 取反 |

### 短路逻辑（这个不只是概念，能救命）

- `and`：如果**左边是 False**，右边**根本不会被执行**（因为结果已经注定是 False）
- `or`：如果**左边是 True**，右边**根本不会被执行**（因为结果已经注定是 True）

这有什么用？看下面的除法例子——它能防止程序崩溃。

````python
x = 0

# 写法一（安全）：先判 x 不为 0，再判除法
if x != 0 and 10 / x > 2:
    print("结果大于 2")
else:
    print("走到了 else —— 因为 x != 0 是 False，"
          "and 发生短路，右边的 10/x 根本没执行，所以没炸")

# 对比：如果把两个条件反过来写
# if 10 / x > 2 and x != 0:     # ← 会 ZeroDivisionError！
#     pass                       #    因为左边先算，10/0 当场崩溃
print()
print("结论：and 连接的多个条件，要把『保护性』的判断放左边，靠短路挡住右边。")
````

````python
# 常规例子
age = 25
has_ticket = True

if age >= 18 and has_ticket:
    print("可以入场")

if age < 12 or age >= 65:
    print("可以享受免票")
else:
    print("需要全票")

# not 取反
is_vip = False
if not is_vip:
    print("普通用户")
````

## 5. `in`：判断「在不在里面」

`in` 能用在三种场景，非常好用：
- 判断**子串**在不在字符串里
- 判断**元素**在不在列表里
- 判断**键**在不在字典里（注意是键，不是值！）

````python
filename = "报关单_2026.pdf"

# 字符串：判断包含
print(f"'pdf' in filename      → {'pdf' in filename}")
print(f"'PDF' in filename      → {'PDF' in filename}   ← 大小写敏感！要判后缀别忘 lower()")
print(f"'报关' in filename     → {'报关' in filename}")

# 列表：判断元素是否存在
allowed = ["admin", "editor", "viewer"]
role = "editor"
print()
print(f"role in allowed        → {role in allowed}")
print(f"'guest' in allowed     → {'guest' in allowed}")

# 字典：in 判断的是【键】，不是值
person = {"name": "张三", "age": 30}
print()
print(f"'name' in person       → {'name' in person}    ← 键存在")
print(f"'张三' in person        → {'张三' in person}   ← 值存在但键不存在！in 不管值")
print(f"'张三' in person.values() → {'张三' in person.values()}   ← 要查值得用 values()")

# 取反：not in
print()
print(f"'excel' not in filename → {'excel' not in filename}")
````

## 6. `is` 与 `==` 的区别

这是你昨天问过的「方括号」之外的另一个经典困惑，也是面试高频题。

- `==` 问的是：**两个东西的内容相等吗？**
- `is` 问的是：**两个名字指的是同一个东西吗？**（是不是同一个对象）

````python
a = [1, 2, 3]
b = [1, 2, 3]      # 内容和 a 一样，但是新做的一份
c = a              # c 和 a 指向同一个对象

print(f"a == b  → {a == b}   内容相同")
print(f"a is b  → {a is b}   不是同一个对象")
print(f"a == c  → {a == c}   内容相同")
print(f"a is c  → {a is c}   是同一个对象")

# 唯一你几乎必须用 is 的场景：判断 None
value = None
if value is None:
    print()
    print("判断『是不是 None』，标准写法是 is None，不是 == None")
print()
print("记不住就先记这条：判断 None 用 is，其它比较一律用 ==。等学到面向对象再深挖。")
````

## 7. 嵌套地狱，与「提前收工」

条件套条件，写多了就没人看得懂——这叫「嵌套地狱」，是烂代码的头号特征。
原则：**能不嵌套就不嵌套，能提前结束就提前结束**。

````python
# ── 嵌套写法（能跑，但难读）──
def check_access_ugly(age, has_ticket, is_vip):
    if age >= 18:
        if has_ticket or is_vip:
            if is_vip:
                return "VIP 通道入场"
            else:
                return "普通入场"
        else:
            return "没票，拒绝"
    else:
        return "未成年，拒绝"

# ── 拍平写法（用『提前 return』把边界情况先挡掉）──
def check_access_clean(age, has_ticket, is_vip):
    if age < 18:
        return "未成年，拒绝"           # 先处理异常，马上收工
    if not has_ticket and not is_vip:
        return "没票，拒绝"             # 再处理下一个异常，收工
    if is_vip:
        return "VIP 通道入场"           # 走到这里只剩正常情况
    return "普通入场"

print(check_access_ugly(20, True, False))
print(check_access_clean(20, True, False))
print(check_access_clean(15, True, False))
print()
print("（函数和 return 下周 Day 8 正式讲，今天先认得这个『先挡异常、再走正路』的思路，")
print("  写多分支时会反复用到。）")
````

```
BMI 公式：体重(公斤) ÷ 身高(米)²

中国成人标准分级：
  < 18.5          偏瘦
  18.5 – 24       正常
  24 – 28         超重

```
